```python
# ==============================================================================
# PHASE 6.5 — CELL 5
# VALIDATE PHASE 6 v4 METADATA SCHEMA — READ ONLY
# ==============================================================================
#
# IMPORTANT:
# - READ ONLY
# - ไม่แก้ clip_metadata.csv
# - ไม่แก้ NPZ
# - ไม่สร้าง dataset
# - ไม่สร้าง label
# - ตรวจ schema ที่ Phase 6 v4 สร้างจริง
#
# Phase 6 v4 schema:
#   clip_id
#   video
#   shard
#   shard_index
#   start_frame
#   end_frame
#
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 78)
print("PHASE 6.5 — CELL 5 VALIDATE PHASE 6 v4 METADATA SCHEMA")
print("=" * 78)


# ==============================================================================
# 1. CONFIG
# ==============================================================================

PROJECT_ROOT = Path(r"C:\LipReadingSSL")
OUTPUT_ROOT = PROJECT_ROOT / "output"

VIDEO_IDS = [
    "video001",
    "video002",
    "video003",
]


# ==============================================================================
# 2. PHASE 6 v4 REQUIRED SCHEMA
# ==============================================================================

# นี่คือ schema ที่ตรงกับ metadata จริงจาก Phase 6 v4
REQUIRED_COLUMNS = [
    "clip_id",
    "video",
    "shard",
    "shard_index",
    "start_frame",
    "end_frame",
]


# ==============================================================================
# 3. OPTIONAL / NOT REQUIRED COLUMNS
# ==============================================================================

# ถ้ามี column อื่นเพิ่มมา ไม่ถือว่า error
OPTIONAL_COLUMNS = [
    "num_frames",
    "video_id",
    "clip_index",
    "shard_id",
    "dtype",
    "height",
    "width",
    "storage",
]


# ==============================================================================
# 4. HELPER
# ==============================================================================

def validate_integer_column(df, column):
    """
    ตรวจว่าคอลัมน์เป็น numeric และไม่มีค่า invalid
    """
    if column not in df.columns:
        return 0

    values = pd.to_numeric(df[column], errors="coerce")

    return int(values.isna().sum())


# ==============================================================================
# 5. VALIDATE EACH VIDEO
# ==============================================================================

all_pass = True

summary_rows = []


for video_id in VIDEO_IDS:

    print()
    print("-" * 78)
    print(video_id)
    print("-" * 78)

    metadata_path = (
        OUTPUT_ROOT
        / video_id
        / "clip_metadata.csv"
    )

    print("Metadata path :", metadata_path)
    print("Exists        :", metadata_path.exists())

    # --------------------------------------------------------------------------
    # File existence
    # --------------------------------------------------------------------------

    if not metadata_path.exists():

        print("❌ Metadata file NOT FOUND")

        all_pass = False

        summary_rows.append({
            "video_id": video_id,
            "rows": 0,
            "missing_columns": "FILE_NOT_FOUND",
            "null_required": -1,
            "invalid_numeric": -1,
            "invalid_frame_range": -1,
            "video_mismatch": -1,
            "duplicate_clip_id": -1,
            "status": "FAIL",
        })

        continue


    # --------------------------------------------------------------------------
    # Load — READ ONLY
    # --------------------------------------------------------------------------

    df = pd.read_csv(metadata_path)

    print("Rows :", f"{len(df):,}")
    print("Columns:")

    for column in df.columns:
        print("  -", column)


    # --------------------------------------------------------------------------
    # Schema validation
    # --------------------------------------------------------------------------

    missing_columns = [
        column
        for column in REQUIRED_COLUMNS
        if column not in df.columns
    ]

    if missing_columns:

        print()
        print("❌ Schema status : FAIL")
        print("Missing required columns :", missing_columns)

        all_pass = False

        summary_rows.append({
            "video_id": video_id,
            "rows": len(df),
            "missing_columns": ",".join(missing_columns),
            "null_required": -1,
            "invalid_numeric": -1,
            "invalid_frame_range": -1,
            "video_mismatch": -1,
            "duplicate_clip_id": -1,
            "status": "FAIL",
        })

        continue


    print("Schema status : PASS")


    # ==============================================================================
    # 6. NULL CHECK
    # ==============================================================================

    null_required = int(
        df[REQUIRED_COLUMNS]
        .isna()
        .any(axis=1)
        .sum()
    )

    print()
    print("Null required rows :", f"{null_required:,}")


    # ==============================================================================
    # 7. NUMERIC CHECK
    # ==============================================================================

    numeric_columns = [
        "shard_index",
        "start_frame",
        "end_frame",
    ]

    invalid_numeric = 0

    numeric_details = {}

    for column in numeric_columns:

        invalid_count = validate_integer_column(
            df,
            column
        )

        numeric_details[column] = invalid_count
        invalid_numeric += invalid_count

    print()
    print("Invalid numeric values :", f"{invalid_numeric:,}")

    for column, count in numeric_details.items():
        print(
            f"  {column:15s}: {count:,}"
        )


    # ==============================================================================
    # 8. FRAME RANGE CHECK
    # ==============================================================================

    start = pd.to_numeric(
        df["start_frame"],
        errors="coerce"
    )

    end = pd.to_numeric(
        df["end_frame"],
        errors="coerce"
    )

    invalid_frame_range = int(
        (
            start.notna()
            & end.notna()
            & (
                (start < 0)
                | (end < 0)
                | (end < start)
            )
        ).sum()
    )

    print()
    print(
        "Invalid frame ranges :",
        f"{invalid_frame_range:,}"
    )


    # ==============================================================================
    # 9. VIDEO CONSISTENCY
    # ==============================================================================

    video_mismatch = 0

    if "video" in df.columns:

        video_values = (
            df["video"]
            .astype(str)
            .str.strip()
        )

        video_mismatch = int(
            (video_values != video_id).sum()
        )

    print()
    print(
        "Video consistency :",
        "PASS" if video_mismatch == 0 else "FAIL"
    )

    if video_mismatch:
        print(
            "  Mismatched rows :",
            f"{video_mismatch:,}"
        )


    # ==============================================================================
    # 10. DUPLICATE CLIP ID
    # ==============================================================================

    duplicate_clip_id = int(
        df["clip_id"]
        .duplicated()
        .sum()
    )

    print()
    print(
        "Duplicate clip_id :",
        f"{duplicate_clip_id:,}"
    )


    # ==============================================================================
    # 11. CLIP ID EMPTY CHECK
    # ==============================================================================

    empty_clip_id = int(
        df["clip_id"]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        "Empty clip_id     :",
        f"{empty_clip_id:,}"
    )


    # ==============================================================================
    # 12. SHARD VALUE CHECK
    # ==============================================================================

    empty_shard = int(
        df["shard"]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print()
    print(
        "Empty shard       :",
        f"{empty_shard:,}"
    )


    # ==============================================================================
    # 13. SHARD INDEX CHECK
    # ==============================================================================

    shard_index = pd.to_numeric(
        df["shard_index"],
        errors="coerce"
    )

    invalid_shard_index = int(
        (
            shard_index.notna()
            & (shard_index < 0)
        ).sum()
    )

    print(
        "Invalid shard_index :",
        f"{invalid_shard_index:,}"
    )


    # ==============================================================================
    # 14. FINAL VIDEO STATUS
    # ==============================================================================

    video_pass = (
        len(missing_columns) == 0
        and null_required == 0
        and invalid_numeric == 0
        and invalid_frame_range == 0
        and video_mismatch == 0
        and duplicate_clip_id == 0
        and empty_clip_id == 0
        and empty_shard == 0
        and invalid_shard_index == 0
    )

    status = "PASS" if video_pass else "FAIL"

    print()
    print(
        f"{'✅' if video_pass else '❌'} {video_id} : {status}"
    )


    if not video_pass:
        all_pass = False


    summary_rows.append({
        "video_id": video_id,
        "rows": len(df),
        "missing_columns": "",
        "null_required": null_required,
        "invalid_numeric": invalid_numeric,
        "invalid_frame_range": invalid_frame_range,
        "video_mismatch": video_mismatch,
        "duplicate_clip_id": duplicate_clip_id,
        "status": status,
    })


# ==============================================================================
# 15. SUMMARY
# ==============================================================================

print()
print("=" * 78)
print("PHASE 6.5 — CELL 5 SUMMARY")
print("=" * 78)

summary_df = pd.DataFrame(summary_rows)

print(
    summary_df.to_string(index=False)
)


# ==============================================================================
# 16. EXPECTED PHASE 6 CLIP COUNTS
# ==============================================================================

expected_counts = {
    "video001": 40076,
    "video002": 31662,
    "video003": 26273,
}

print()
print("=" * 78)
print("PHASE 6 v4 EXPECTED CLIP COUNTS")
print("=" * 78)

count_check_pass = True

for video_id, expected in expected_counts.items():

    row = summary_df[
        summary_df["video_id"] == video_id
    ]

    if row.empty:
        print(
            f"❌ {video_id}: metadata missing"
        )
        count_check_pass = False
        continue

    actual = int(row.iloc[0]["rows"])

    status = "PASS" if actual == expected else "FAIL"

    print(
        f"{video_id}: "
        f"actual={actual:,} | "
        f"expected={expected:,} | "
        f"{status}"
    )

    if actual != expected:
        count_check_pass = False


# ==============================================================================
# 17. FINAL RESULT
# ==============================================================================

print()
print("=" * 78)
print("CELL 5 FINAL RESULT")
print("=" * 78)

if all_pass and count_check_pass:

    print("✅ CELL 5 READ-ONLY VALIDATION : PASS")
    print()
    print("Phase 6 v4 metadata schema is valid.")
    print("All three videos are present.")
    print("Clip counts match Phase 6 v4 expected counts.")
    print()
    print("Ready for next Phase 6.5 cell.")

else:

    print("❌ CELL 5 READ-ONLY VALIDATION : FAIL")
    print()
    print("Do NOT modify Phase 6 files yet.")
    print("Inspect the reported issue before continuing.")


print()
print("=" * 78)
print("CELL 5 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ============================================================
# PHASE 6.5 — CELL 6
# VALIDATE GLOBAL CLIP ID UNIQUENESS
#
# READ ONLY
# ============================================================

print("=" * 78)
print("PHASE 6.5 — CELL 6")
print("VALIDATE GLOBAL CLIP ID UNIQUENESS")
print("=" * 78)


global_duplicate_results = []


for video_id in VIDEO_IDS:

    df = PHASE65_METADATA[
        video_id
    ].copy()

    duplicate_count = int(
        df[
            "clip_id"
        ]
        .duplicated()
        .sum()
    )


    unique_count = int(
        df[
            "clip_id"
        ]
        .nunique()
    )


    total_count = len(df)


    passed = (
        duplicate_count == 0
        and unique_count == total_count
    )


    global_duplicate_results.append({
        "video_id": video_id,
        "rows": total_count,
        "unique_clip_ids": unique_count,
        "duplicate_clip_ids": duplicate_count,
        "status": "PASS" if passed else "FAIL",
    })


    print()
    print(
        f"{video_id}"
    )

    print(
        "Rows              :",
        f"{total_count:,}"
    )

    print(
        "Unique clip IDs   :",
        f"{unique_count:,}"
    )

    print(
        "Duplicate IDs     :",
        f"{duplicate_count:,}"
    )

    print(
        "Status            :",
        "PASS" if passed else "FAIL"
    )


duplicate_df = pd.DataFrame(
    global_duplicate_results
)


if (
    duplicate_df[
        "status"
    ]
    == "FAIL"
).any():

    raise RuntimeError(
        "Duplicate global clip IDs detected "
        "in Phase 6 metadata."
    )


print()
print("=" * 78)
print("CELL 6 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 7
# VALIDATE FRAME RANGES — PHASE 6 v4
# READ ONLY
# ==============================================================================
#
# Phase 6 v4 metadata schema:
#   clip_id
#   video
#   shard
#   shard_index
#   start_frame
#   end_frame
#
# IMPORTANT:
# - ไม่มีการอ่าน num_frames จาก CSV
# - คำนวณ num_frames จาก start_frame/end_frame
# - ไม่แก้ไขไฟล์ใด ๆ
# - ไม่สร้างไฟล์ใด ๆ
#
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 78)
print("PHASE 6.5 — CELL 7")
print("VALIDATE FRAME RANGES — PHASE 6 v4")
print("=" * 78)


# ==============================================================================
# 1. CONFIG
# ==============================================================================

PROJECT_ROOT = Path(r"C:\LipReadingSSL")
OUTPUT_ROOT = PROJECT_ROOT / "output"

VIDEO_IDS = [
    "video001",
    "video002",
    "video003",
]


# ==============================================================================
# 2. EXPECTED CLIP COUNTS
# ==============================================================================

EXPECTED_CLIPS = {
    "video001": 40076,
    "video002": 31662,
    "video003": 26273,
}


# ==============================================================================
# 3. REQUIRED PHASE 6 v4 COLUMNS
# ==============================================================================

REQUIRED_COLUMNS = [
    "clip_id",
    "video",
    "shard",
    "shard_index",
    "start_frame",
    "end_frame",
]


# ==============================================================================
# 4. VALIDATION RESULTS
# ==============================================================================

results = []

all_pass = True


# ==============================================================================
# 5. PROCESS EACH VIDEO
# ==============================================================================

for video_id in VIDEO_IDS:

    print()
    print("-" * 78)
    print(video_id)
    print("-" * 78)

    metadata_path = (
        OUTPUT_ROOT
        / video_id
        / "clip_metadata.csv"
    )

    print("Metadata :", metadata_path)
    print("Exists   :", metadata_path.exists())

    # --------------------------------------------------------------------------
    # File check
    # --------------------------------------------------------------------------

    if not metadata_path.exists():

        print("❌ Metadata file not found")

        results.append({
            "video_id": video_id,
            "rows": 0,
            "negative_start": -1,
            "negative_end": -1,
            "end_before_start": -1,
            "invalid_numeric": -1,
            "duplicate_clip_id": -1,
            "video_mismatch": -1,
            "status": "FAIL",
        })

        all_pass = False

        continue


    # --------------------------------------------------------------------------
    # READ ONLY
    # --------------------------------------------------------------------------

    df = pd.read_csv(metadata_path)


    # --------------------------------------------------------------------------
    # Schema check
    # --------------------------------------------------------------------------

    missing = [
        col
        for col in REQUIRED_COLUMNS
        if col not in df.columns
    ]

    if missing:

        print("❌ Missing required columns:", missing)

        results.append({
            "video_id": video_id,
            "rows": len(df),
            "negative_start": -1,
            "negative_end": -1,
            "end_before_start": -1,
            "invalid_numeric": -1,
            "duplicate_clip_id": -1,
            "video_mismatch": -1,
            "status": "FAIL",
        })

        all_pass = False

        continue


    print("Schema : PASS")


    # ==============================================================================
    # 6. CONVERT FRAME COLUMNS TO NUMERIC
    # ==============================================================================

    start = pd.to_numeric(
        df["start_frame"],
        errors="coerce"
    )

    end = pd.to_numeric(
        df["end_frame"],
        errors="coerce"
    )


    # ==============================================================================
    # 7. INVALID NUMERIC VALUES
    # ==============================================================================

    invalid_numeric = int(
        start.isna().sum()
        +
        end.isna().sum()
    )

    print()
    print(
        "Invalid numeric frame values :",
        f"{invalid_numeric:,}"
    )


    # ==============================================================================
    # 8. NEGATIVE FRAME CHECK
    # ==============================================================================

    negative_start = int(
        (start.notna() & (start < 0)).sum()
    )

    negative_end = int(
        (end.notna() & (end < 0)).sum()
    )

    print(
        "Negative start_frame         :",
        f"{negative_start:,}"
    )

    print(
        "Negative end_frame           :",
        f"{negative_end:,}"
    )


    # ==============================================================================
    # 9. FRAME ORDER CHECK
    # ==============================================================================

    end_before_start = int(
        (
            start.notna()
            &
            end.notna()
            &
            (end < start)
        ).sum()
    )

    print(
        "end_frame < start_frame      :",
        f"{end_before_start:,}"
    )


    # ==============================================================================
    # 10. CALCULATE num_frames
    # ==============================================================================
    #
    # Phase 6 v4 ไม่มี num_frames column
    #
    # ดังนั้น:
    #
    #     num_frames = end_frame - start_frame + 1
    #
    # เช่น:
    #     start=0
    #     end=15
    #
    #     num_frames = 15 - 0 + 1 = 16
    #
    # ==============================================================================

    calculated_num_frames = (
        end - start + 1
    )


    # ==============================================================================
    # 11. INVALID CALCULATED FRAME COUNT
    # ==============================================================================

    invalid_num_frames = int(
        (
            calculated_num_frames.notna()
            &
            (calculated_num_frames <= 0)
        ).sum()
    )

    print(
        "Invalid calculated num_frames :",
        f"{invalid_num_frames:,}"
    )


    # ==============================================================================
    # 12. DUPLICATE CLIP ID
    # ==============================================================================

    duplicate_clip_id = int(
        df["clip_id"]
        .astype(str)
        .str.strip()
        .duplicated()
        .sum()
    )

    print(
        "Duplicate clip_id             :",
        f"{duplicate_clip_id:,}"
    )


    # ==============================================================================
    # 13. VIDEO CONSISTENCY
    # ==============================================================================

    video_values = (
        df["video"]
        .astype(str)
        .str.strip()
    )

    video_mismatch = int(
        (video_values != video_id).sum()
    )

    print(
        "Video mismatch                 :",
        f"{video_mismatch:,}"
    )


    # ==============================================================================
    # 14. CLIP COUNT CHECK
    # ==============================================================================

    actual_count = len(df)

    expected_count = EXPECTED_CLIPS[video_id]

    clip_count_pass = (
        actual_count == expected_count
    )

    print()
    print(
        "Clip count :",
        f"{actual_count:,}",
        "/ expected",
        f"{expected_count:,}",
        ":",
        "PASS" if clip_count_pass else "FAIL"
    )


    # ==============================================================================
    # 15. FINAL VIDEO VALIDATION
    # ==============================================================================

    video_pass = (
        invalid_numeric == 0
        and negative_start == 0
        and negative_end == 0
        and end_before_start == 0
        and invalid_num_frames == 0
        and duplicate_clip_id == 0
        and video_mismatch == 0
        and clip_count_pass
    )

    status = "PASS" if video_pass else "FAIL"

    print()
    print(
        f"{'✅' if video_pass else '❌'} "
        f"{video_id} FRAME RANGE : {status}"
    )


    if not video_pass:
        all_pass = False


    results.append({
        "video_id": video_id,
        "rows": actual_count,
        "negative_start": negative_start,
        "negative_end": negative_end,
        "end_before_start": end_before_start,
        "invalid_numeric": invalid_numeric,
        "invalid_num_frames": invalid_num_frames,
        "duplicate_clip_id": duplicate_clip_id,
        "video_mismatch": video_mismatch,
        "status": status,
    })


# ==============================================================================
# 16. SUMMARY
# ==============================================================================

print()
print("=" * 78)
print("PHASE 6.5 — CELL 7 SUMMARY")
print("=" * 78)

results_df = pd.DataFrame(results)

print(
    results_df.to_string(index=False)
)


# ==============================================================================
# 17. TOTAL DATASET CHECK
# ==============================================================================

print()
print("=" * 78)
print("TOTAL PHASE 6 v4 DATASET")
print("=" * 78)

actual_total = int(
    results_df["rows"].sum()
)

expected_total = sum(
    EXPECTED_CLIPS.values()
)

print(
    "Actual total   :",
    f"{actual_total:,}"
)

print(
    "Expected total :",
    f"{expected_total:,}"
)

total_pass = (
    actual_total == expected_total
)

print(
    "Total count    :",
    "PASS" if total_pass else "FAIL"
)


# ==============================================================================
# 18. FINAL RESULT
# ==============================================================================

print()
print("=" * 78)
print("CELL 7 FINAL RESULT")
print("=" * 78)

if all_pass and total_pass:

    print("✅ CELL 7 FRAME RANGE VALIDATION : PASS")
    print()
    print("All three videos passed.")
    print("All frame ranges are valid.")
    print("num_frames was calculated from start_frame/end_frame.")
    print("No num_frames column is required.")
    print()
    print(
        f"Total clips : {actual_total:,}"
    )

else:

    print("❌ CELL 7 FRAME RANGE VALIDATION : FAIL")
    print()
    print("Review the validation summary above.")
    print("No files were modified.")


print()
print("=" * 78)
print("CELL 7 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 8
# VALIDATE METADATA SHARD MAPPING — PHASE 6 v4
# READ ONLY
# ==============================================================================
#
# Phase 6 v4 metadata schema:
#
#   clip_id
#   video
#   shard
#   shard_index
#   start_frame
#   end_frame
#
# IMPORTANT:
# - ไม่มี shard_id
# - ใช้ shard
# - ใช้ shard_index
# - ไม่แก้ไข metadata
# - ไม่แก้ไข NPZ
# - ไม่สร้างไฟล์
#
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 78)
print("PHASE 6.5 — CELL 8")
print("VALIDATE METADATA SHARD MAPPING — PHASE 6 v4")
print("=" * 78)


# ==============================================================================
# 1. CONFIG
# ==============================================================================

PROJECT_ROOT = Path(r"C:\LipReadingSSL")
OUTPUT_ROOT = PROJECT_ROOT / "output"

VIDEO_IDS = [
    "video001",
    "video002",
    "video003",
]


# ==============================================================================
# 2. EXPECTED CLIP COUNTS
# ==============================================================================

EXPECTED_CLIPS = {
    "video001": 40076,
    "video002": 31662,
    "video003": 26273,
}


# ==============================================================================
# 3. REQUIRED PHASE 6 v4 COLUMNS
# ==============================================================================

REQUIRED_COLUMNS = [
    "clip_id",
    "video",
    "shard",
    "shard_index",
    "start_frame",
    "end_frame",
]


# ==============================================================================
# 4. RESULT STORAGE
# ==============================================================================

mapping_results = []

all_pass = True


# ==============================================================================
# 5. PROCESS EACH VIDEO
# ==============================================================================

for video_id in VIDEO_IDS:

    print()
    print("-" * 78)
    print(video_id)
    print("-" * 78)

    video_root = (
        OUTPUT_ROOT
        / video_id
    )

    metadata_path = (
        video_root
        / "clip_metadata.csv"
    )

    shard_dir = (
        video_root
        / "clip_npy"
    )

    print("Metadata :", metadata_path)
    print("Shards   :", shard_dir)

    # ==========================================================================
    # FILE CHECKS
    # ==========================================================================

    if not metadata_path.exists():

        print("❌ Metadata file not found")

        mapping_results.append({
            "video_id": video_id,
            "metadata_rows": 0,
            "shards": 0,
            "invalid_shard": -1,
            "invalid_shard_index": -1,
            "missing_shard_file": -1,
            "shard_index_out_of_range": -1,
            "duplicate_mapping": -1,
            "status": "FAIL",
        })

        all_pass = False
        continue


    if not shard_dir.exists():

        print("❌ Shard directory not found")

        mapping_results.append({
            "video_id": video_id,
            "metadata_rows": 0,
            "shards": 0,
            "invalid_shard": -1,
            "invalid_shard_index": -1,
            "missing_shard_file": -1,
            "shard_index_out_of_range": -1,
            "duplicate_mapping": -1,
            "status": "FAIL",
        })

        all_pass = False
        continue


    # ==========================================================================
    # READ METADATA — READ ONLY
    # ==========================================================================

    df = pd.read_csv(metadata_path)

    print(
        "Metadata rows :",
        f"{len(df):,}"
    )

    print(
        "Expected rows :",
        f"{EXPECTED_CLIPS[video_id]:,}"
    )


    # ==========================================================================
    # SCHEMA CHECK
    # ==========================================================================

    missing = [
        col
        for col in REQUIRED_COLUMNS
        if col not in df.columns
    ]

    if missing:

        print(
            "❌ Missing required columns:",
            missing
        )

        mapping_results.append({
            "video_id": video_id,
            "metadata_rows": len(df),
            "shards": 0,
            "invalid_shard": -1,
            "invalid_shard_index": -1,
            "missing_shard_file": -1,
            "shard_index_out_of_range": -1,
            "duplicate_mapping": -1,
            "status": "FAIL",
        })

        all_pass = False
        continue


    print("Schema : PASS")


    # ==========================================================================
    # 6. LIST ACTUAL SHARD FILES
    # ==========================================================================

    shard_files = sorted(
        shard_dir.glob("*.npz")
    )

    print(
        "Shard files :",
        f"{len(shard_files):,}"
    )

    if len(shard_files) == 0:

        print("❌ No NPZ shard files found")

        mapping_results.append({
            "video_id": video_id,
            "metadata_rows": len(df),
            "shards": 0,
            "invalid_shard": -1,
            "invalid_shard_index": -1,
            "missing_shard_file": -1,
            "shard_index_out_of_range": -1,
            "duplicate_mapping": -1,
            "status": "FAIL",
        })

        all_pass = False
        continue


    # ==========================================================================
    # 7. NORMALIZE SHARD COLUMN
    # ==========================================================================

    shard_values = (
        df["shard"]
        .astype(str)
        .str.strip()
    )


    # ==========================================================================
    # 8. VALIDATE SHARD NAMES
    # ==========================================================================

    actual_shard_names = {
        p.name
        for p in shard_files
    }

    invalid_shard_mask = (
        ~shard_values.isin(actual_shard_names)
    )

    invalid_shard = int(
        invalid_shard_mask.sum()
    )

    print(
        "Invalid shard names :",
        f"{invalid_shard:,}"
    )


    # ==========================================================================
    # 9. NORMALIZE SHARD INDEX
    # ==========================================================================

    shard_index = pd.to_numeric(
        df["shard_index"],
        errors="coerce"
    )


    invalid_shard_index = int(
        shard_index.isna().sum()
    )

    print(
        "Invalid shard_index :",
        f"{invalid_shard_index:,}"
    )


    # ==========================================================================
    # 10. CACHE SHARD CAPACITY
    # ==========================================================================
    #
    # We inspect each NPZ only for validation.
    #
    # No file is modified.
    #
    # The validator attempts to determine the number of clips stored
    # in each shard.
    #
    # ==========================================================================

    shard_capacity = {}

    shard_load_errors = []

    print()
    print("Reading shard capacities...")


    for i, shard_path in enumerate(shard_files, start=1):

        try:

            with np.load(
                shard_path,
                allow_pickle=False
            ) as data:

                # --------------------------------------------------------------
                # Find the clip/frame array.
                #
                # Phase 6 v4 shards are expected to contain the actual clips.
                # We identify the largest array whose first dimension is usable.
                # --------------------------------------------------------------

                candidates = []

                for key in data.files:

                    arr = data[key]

                    if getattr(arr, "ndim", 0) >= 1:

                        candidates.append(
                            (
                                key,
                                int(arr.shape[0])
                            )
                        )

                if not candidates:

                    raise RuntimeError(
                        "No usable array found in NPZ"
                    )


                # Prefer common clip keys if present.

                preferred_keys = [
                    "clips",
                    "frames",
                    "data",
                    "mouth",
                    "video",
                    "x",
                ]

                selected = None

                for preferred in preferred_keys:

                    for key, size in candidates:

                        if key.lower() == preferred.lower():

                            selected = size
                            break

                    if selected is not None:
                        break


                # Fallback:
                # use the largest first dimension.

                if selected is None:

                    selected = max(
                        size
                        for _, size in candidates
                    )


                shard_capacity[
                    shard_path.name
                ] = int(selected)


        except Exception as exc:

            shard_load_errors.append(
                (
                    shard_path.name,
                    str(exc)
                )
            )


    print(
        "Shard capacity records :",
        f"{len(shard_capacity):,}"
    )


    if shard_load_errors:

        print()
        print(
            "❌ Shard read errors :",
            f"{len(shard_load_errors):,}"
        )

        for shard_name, error in shard_load_errors[:10]:

            print(
                " ",
                shard_name,
                "->",
                error
            )


    # ==========================================================================
    # 11. CHECK MISSING SHARD FILE
    # ==========================================================================

    missing_shard_file = int(
        (
            ~shard_values.isin(shard_capacity.keys())
        ).sum()
    )

    print()
    print(
        "Missing / unreadable shard :",
        f"{missing_shard_file:,}"
    )


    # ==========================================================================
    # 12. CHECK SHARD INDEX RANGE
    # ==========================================================================

    out_of_range = 0

    valid_capacity_mask = shard_values.isin(
        shard_capacity.keys()
    )

    valid_index_mask = shard_index.notna()

    rows_to_check = (
        valid_capacity_mask
        &
        valid_index_mask
    )


    if rows_to_check.any():

        capacities = shard_values.map(
            shard_capacity
        )

        out_of_range_mask = (
            rows_to_check
            &
            (
                (shard_index < 0)
                |
                (shard_index >= capacities)
            )
        )

        out_of_range = int(
            out_of_range_mask.sum()
        )


    print(
        "Shard index out of range :",
        f"{out_of_range:,}"
    )


    # ==========================================================================
    # 13. CHECK DUPLICATE SHARD MAPPING
    # ==========================================================================
    #
    # One metadata row should correspond to one:
    #
    #     shard + shard_index
    #
    # pair.
    #
    # ==========================================================================

    mapping_key = (
        shard_values
        + "::"
        + shard_index.astype("Int64")
        .astype(str)
    )

    duplicate_mapping = int(
        mapping_key.duplicated().sum()
    )

    print(
        "Duplicate shard mapping :",
        f"{duplicate_mapping:,}"
    )


    # ==========================================================================
    # 14. CLIP COUNT CHECK
    # ==========================================================================

    actual_rows = len(df)

    expected_rows = EXPECTED_CLIPS[video_id]

    clip_count_pass = (
        actual_rows == expected_rows
    )


    # ==========================================================================
    # 15. FINAL VIDEO RESULT
    # ==========================================================================

    video_pass = (
        clip_count_pass
        and invalid_shard == 0
        and invalid_shard_index == 0
        and missing_shard_file == 0
        and out_of_range == 0
        and duplicate_mapping == 0
        and len(shard_load_errors) == 0
    )

    status = (
        "PASS"
        if video_pass
        else
        "FAIL"
    )


    print()
    print(
        f"{'✅' if video_pass else '❌'} "
        f"{video_id} SHARD MAPPING : {status}"
    )


    if not video_pass:
        all_pass = False


    mapping_results.append({
        "video_id": video_id,
        "metadata_rows": actual_rows,
        "shards": len(shard_files),
        "invalid_shard": invalid_shard,
        "invalid_shard_index": invalid_shard_index,
        "missing_shard_file": missing_shard_file,
        "shard_index_out_of_range": out_of_range,
        "duplicate_mapping": duplicate_mapping,
        "status": status,
    })


# ==============================================================================
# 16. SUMMARY
# ==============================================================================

print()
print("=" * 78)
print("PHASE 6.5 — CELL 8 SUMMARY")
print("=" * 78)

mapping_results_df = pd.DataFrame(
    mapping_results
)

print(
    mapping_results_df.to_string(
        index=False
    )
)


# ==============================================================================
# 17. EXPECTED TOTAL
# ==============================================================================

actual_total = int(
    mapping_results_df["metadata_rows"]
    .sum()
)

expected_total = sum(
    EXPECTED_CLIPS.values()
)

print()
print("=" * 78)
print("TOTAL METADATA")
print("=" * 78)

print(
    "Actual total   :",
    f"{actual_total:,}"
)

print(
    "Expected total :",
    f"{expected_total:,}"
)

print(
    "Total count    :",
    "PASS"
    if actual_total == expected_total
    else
    "FAIL"
)


# ==============================================================================
# 18. FINAL RESULT
# ==============================================================================

print()
print("=" * 78)
print("CELL 8 FINAL RESULT")
print("=" * 78)

if (
    all_pass
    and actual_total == expected_total
):

    print("✅ CELL 8 SHARD MAPPING VALIDATION : PASS")
    print()
    print("All three videos passed.")
    print("Metadata → shard → shard_index mapping is valid.")
    print()
    print(
        f"Total clips : {actual_total:,}"
    )

else:

    print("❌ CELL 8 SHARD MAPPING VALIDATION : FAIL")
    print()
    print("Review the summary above.")
    print("No files were modified.")


print()
print("=" * 78)
print("CELL 8 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 9
# VALIDATE EVERY NPZ SHARD — PHASE 6 v4
# READ ONLY
# ==============================================================================
#
# Purpose:
#   ตรวจสอบ NPZ ทุก shard ของ video001, video002, video003
#   ว่า:
#
#   1. ไฟล์ NPZ เปิดได้
#   2. มีข้อมูลภายใน
#   3. จำนวน clip ใน NPZ สอดคล้องกับ metadata
#   4. shard_index ใน metadata อยู่ในขอบเขตของ NPZ
#   5. ไม่มี mapping ซ้ำ
#
# Phase 6 v4 metadata schema:
#
#   clip_id
#   video
#   shard
#   shard_index
#   start_frame
#   end_frame
#
# IMPORTANT:
#   - ไม่มี shard_id
#   - ใช้ shard
#   - ใช้ shard_index
#   - READ ONLY
#   - ไม่แก้ metadata
#   - ไม่แก้ NPZ
#   - ไม่สร้าง dataset ใหม่
#
# ==============================================================================

from pathlib import Path
import pandas as pd
import numpy as np


print("=" * 78)
print("PHASE 6.5 — CELL 9")
print("VALIDATE EVERY NPZ SHARD — PHASE 6 v4")
print("=" * 78)


# ==============================================================================
# 1. CONFIGURATION
# ==============================================================================

PROJECT_ROOT = Path(
    r"C:\LipReadingSSL"
)

OUTPUT_ROOT = (
    PROJECT_ROOT / "output"
)

VIDEO_IDS = [
    "video001",
    "video002",
    "video003",
]


# ==============================================================================
# 2. EXPECTED CLIP COUNTS
# ==============================================================================

EXPECTED_CLIPS = {
    "video001": 40076,
    "video002": 31662,
    "video003": 26273,
}


# ==============================================================================
# 3. REQUIRED METADATA SCHEMA
# ==============================================================================

REQUIRED_COLUMNS = [
    "clip_id",
    "video",
    "shard",
    "shard_index",
    "start_frame",
    "end_frame",
]


# ==============================================================================
# 4. HELPER
# ==============================================================================

def detect_clip_count_from_npz(npz_path):
    """
    READ ONLY

    ตรวจจำนวนรายการที่เก็บอยู่ใน NPZ

    Phase 6 v4 อาจใช้ชื่อ array ต่างกันได้
    จึงตรวจจาก first dimension ของ array ที่เหมาะสม
    """

    with np.load(
        npz_path,
        allow_pickle=False
    ) as data:

        keys = list(data.files)

        if len(keys) == 0:

            raise RuntimeError(
                "NPZ contains no arrays"
            )

        # --------------------------------------------------------------
        # Preferred clip-array names
        # --------------------------------------------------------------

        preferred_keys = [
            "clips",
            "frames",
            "data",
            "mouth",
            "video",
            "x",
        ]

        for preferred in preferred_keys:

            for key in keys:

                if key.lower() == preferred.lower():

                    arr = data[key]

                    if getattr(arr, "ndim", 0) >= 1:

                        return {
                            "count": int(arr.shape[0]),
                            "key": key,
                            "keys": keys,
                        }


        # --------------------------------------------------------------
        # Fallback
        # --------------------------------------------------------------

        candidates = []

        for key in keys:

            arr = data[key]

            if getattr(arr, "ndim", 0) >= 1:

                candidates.append(
                    (
                        key,
                        int(arr.shape[0]),
                        tuple(arr.shape),
                    )
                )

        if not candidates:

            raise RuntimeError(
                "No usable array with first dimension found"
            )

        # ใช้ array ที่มี first dimension มากที่สุด
        selected = max(
            candidates,
            key=lambda x: x[1]
        )

        return {
            "count": selected[1],
            "key": selected[0],
            "keys": keys,
        }


# ==============================================================================
# 5. STORAGE
# ==============================================================================

all_results = []

global_pass = True


# ==============================================================================
# 6. PROCESS EACH VIDEO
# ==============================================================================

for video_id in VIDEO_IDS:

    print()
    print("=" * 78)
    print(f"VIDEO : {video_id}")
    print("=" * 78)

    metadata_path = (
        OUTPUT_ROOT
        / video_id
        / "clip_metadata.csv"
    )

    shard_dir = (
        OUTPUT_ROOT
        / video_id
        / "clip_npy"
    )


    print(
        "Metadata :",
        metadata_path
    )

    print(
        "Shard dir:",
        shard_dir
    )


    # ==========================================================================
    # 6.1 FILE CHECK
    # ==========================================================================

    if not metadata_path.exists():

        print("❌ Metadata not found")

        global_pass = False

        continue


    if not shard_dir.exists():

        print("❌ Shard directory not found")

        global_pass = False

        continue


    # ==========================================================================
    # 6.2 READ METADATA
    # ==========================================================================

    df = pd.read_csv(
        metadata_path
    )

    print(
        "Metadata rows :",
        f"{len(df):,}"
    )


    # ==========================================================================
    # 6.3 SCHEMA CHECK
    # ==========================================================================

    missing_columns = [
        col
        for col in REQUIRED_COLUMNS
        if col not in df.columns
    ]


    if missing_columns:

        print(
            "❌ Missing columns:",
            missing_columns
        )

        global_pass = False

        continue


    print(
        "Schema : PASS"
    )


    # ==========================================================================
    # 6.4 NORMALIZE SHARD COLUMN
    # ==========================================================================

    metadata_shard = (
        df["shard"]
        .astype(str)
        .str.strip()
    )


    # ==========================================================================
    # 6.5 NORMALIZE SHARD INDEX
    # ==========================================================================

    metadata_shard_index = pd.to_numeric(
        df["shard_index"],
        errors="coerce"
    )


    invalid_shard_index = int(
        metadata_shard_index.isna().sum()
    )


    if invalid_shard_index:

        print(
            "❌ Invalid shard_index:",
            f"{invalid_shard_index:,}"
        )

        global_pass = False

    else:

        print(
            "shard_index : PASS"
        )


    # ==========================================================================
    # 6.6 LIST NPZ FILES
    # ==========================================================================

    shard_files = sorted(
        shard_dir.glob("*.npz")
    )


    print(
        "NPZ shards :",
        f"{len(shard_files):,}"
    )


    if len(shard_files) == 0:

        print(
            "❌ No NPZ files found"
        )

        global_pass = False

        continue


    actual_shard_names = {
        p.name
        for p in shard_files
    }


    # ==========================================================================
    # 6.7 METADATA → NPZ EXISTENCE CHECK
    # ==========================================================================

    missing_shards = sorted(
        set(metadata_shard)
        - actual_shard_names
    )


    if missing_shards:

        print(
            "❌ Metadata references missing NPZ shards:",
            len(missing_shards)
        )

        for name in missing_shards[:10]:

            print(
                "   ",
                name
            )

        global_pass = False

    else:

        print(
            "Metadata → NPZ existence : PASS"
        )


    # ==========================================================================
    # 6.8 VALIDATE EVERY NPZ
    # ==========================================================================

    print()
    print(
        "Validating NPZ files..."
    )


    shard_info = {}

    shard_errors = []


    for index, shard_path in enumerate(
        shard_files,
        start=1
    ):

        try:

            info = detect_clip_count_from_npz(
                shard_path
            )

            shard_info[
                shard_path.name
            ] = info


            print(
                f"[{index:>3}/{len(shard_files):<3}] "
                f"{shard_path.name:<24} "
                f"clips={info['count']:,} "
                f"array={info['key']}"
            )


        except Exception as exc:

            shard_errors.append(
                (
                    shard_path.name,
                    str(exc)
                )
            )


            print(
                f"[{index:>3}/{len(shard_files):<3}] "
                f"{shard_path.name:<24} "
                f"❌ ERROR"
            )


    # ==========================================================================
    # 6.9 NPZ OPEN CHECK
    # ==========================================================================

    if shard_errors:

        print()
        print(
            "❌ NPZ open/read errors :",
            f"{len(shard_errors):,}"
        )

        for shard_name, error in shard_errors[:10]:

            print(
                "   ",
                shard_name,
                "->",
                error
            )

        global_pass = False

    else:

        print()
        print(
            "NPZ open/read : PASS"
        )


    # ==========================================================================
    # 6.10 METADATA SHARD INDEX RANGE
    # ==========================================================================

    out_of_range = 0


    for shard_name, capacity_info in shard_info.items():

        capacity = capacity_info["count"]


        mask = (
            metadata_shard
            == shard_name
        )


        if not mask.any():
            continue


        indices = (
            metadata_shard_index[mask]
        )


        bad = (
            (indices < 0)
            |
            (indices >= capacity)
        )


        out_of_range += int(
            bad.sum()
        )


    print(
        "shard_index range :",
        "PASS"
        if out_of_range == 0
        else f"FAIL ({out_of_range:,})"
    )


    if out_of_range:

        global_pass = False


    # ==========================================================================
    # 6.11 DUPLICATE SHARD + INDEX MAPPING
    # ==========================================================================

    mapping_df = pd.DataFrame({
        "shard": metadata_shard,
        "shard_index": metadata_shard_index,
    })


    duplicate_mapping = int(
        mapping_df.duplicated(
            subset=[
                "shard",
                "shard_index",
            ]
        ).sum()
    )


    print(
        "Duplicate shard mapping :",
        "PASS"
        if duplicate_mapping == 0
        else f"FAIL ({duplicate_mapping:,})"
    )


    if duplicate_mapping:

        global_pass = False


    # ==========================================================================
    # 6.12 METADATA ROW COUNT
    # ==========================================================================

    metadata_count = len(df)

    expected_count = (
        EXPECTED_CLIPS[video_id]
    )


    count_pass = (
        metadata_count
        == expected_count
    )


    print(
        "Metadata clip count :",
        "PASS"
        if count_pass
        else
        (
            f"FAIL "
            f"(actual={metadata_count:,}, "
            f"expected={expected_count:,})"
        )
    )


    if not count_pass:

        global_pass = False


    # ==========================================================================
    # 6.13 SUM NPZ CLIP COUNT
    # ==========================================================================

    npz_total = sum(
        info["count"]
        for info in shard_info.values()
    )


    npz_count_pass = (
        npz_total
        == expected_count
    )


    print(
        "NPZ total clip count :",
        "PASS"
        if npz_count_pass
        else
        (
            f"FAIL "
            f"(actual={npz_total:,}, "
            f"expected={expected_count:,})"
        )
    )


    if not npz_count_pass:

        global_pass = False


    # ==========================================================================
    # 6.14 METADATA → NPZ COUNT CONSISTENCY
    # ==========================================================================

    metadata_mapping_counts = (
        metadata_shard
        .value_counts()
        .to_dict()
    )


    count_mismatches = []


    for shard_name, info in shard_info.items():

        metadata_n = int(
            metadata_mapping_counts.get(
                shard_name,
                0
            )
        )

        npz_n = int(
            info["count"]
        )


        if metadata_n != npz_n:

            count_mismatches.append({
                "shard": shard_name,
                "metadata": metadata_n,
                "npz": npz_n,
            })


    print(
        "Metadata ↔ NPZ per-shard count :",
        "PASS"
        if not count_mismatches
        else
        f"FAIL ({len(count_mismatches):,} shards)"
    )


    if count_mismatches:

        global_pass = False

        for item in count_mismatches[:10]:

            print(
                "   ",
                item
            )


    # ==========================================================================
    # 6.15 VIDEO CONSISTENCY
    # ==========================================================================

    video_values = (
        df["video"]
        .astype(str)
        .str.strip()
    )


    wrong_video = int(
        (
            video_values
            != video_id
        ).sum()
    )


    print(
        "Video consistency :",
        "PASS"
        if wrong_video == 0
        else f"FAIL ({wrong_video:,})"
    )


    if wrong_video:

        global_pass = False


    # ==========================================================================
    # 6.16 CLIP ID UNIQUENESS
    # ==========================================================================

    duplicate_clip_ids = int(
        df["clip_id"]
        .duplicated()
        .sum()
    )


    print(
        "Clip ID uniqueness :",
        "PASS"
        if duplicate_clip_ids == 0
        else f"FAIL ({duplicate_clip_ids:,})"
    )


    if duplicate_clip_ids:

        global_pass = False


    # ==========================================================================
    # 6.17 VIDEO RESULT
    # ==========================================================================

    video_pass = (
        count_pass
        and npz_count_pass
        and not shard_errors
        and out_of_range == 0
        and duplicate_mapping == 0
        and not count_mismatches
        and wrong_video == 0
        and duplicate_clip_ids == 0
        and invalid_shard_index == 0
        and not missing_shards
    )


    print()
    print(
        "=" * 78
    )

    if video_pass:

        print(
            f"✅ {video_id} : PASS"
        )

    else:

        print(
            f"❌ {video_id} : FAIL"
        )

        global_pass = False


    print(
        "=" * 78
    )


    # ==========================================================================
    # 6.18 STORE RESULT
    # ==========================================================================

    all_results.append({
        "video_id": video_id,
        "metadata_rows": metadata_count,
        "expected_rows": expected_count,
        "npz_shards": len(shard_files),
        "npz_total_clips": npz_total,
        "missing_shards": len(missing_shards),
        "npz_read_errors": len(shard_errors),
        "invalid_shard_index": invalid_shard_index,
        "out_of_range": out_of_range,
        "duplicate_mapping": duplicate_mapping,
        "count_mismatches": len(count_mismatches),
        "wrong_video": wrong_video,
        "duplicate_clip_ids": duplicate_clip_ids,
        "status": (
            "PASS"
            if video_pass
            else "FAIL"
        ),
    })


# ==============================================================================
# 7. GLOBAL SUMMARY
# ==============================================================================

print()
print("=" * 78)
print("PHASE 6.5 — CELL 9 GLOBAL SUMMARY")
print("=" * 78)


results_df = pd.DataFrame(
    all_results
)


if len(results_df):

    print(
        results_df.to_string(
            index=False
        )
    )


# ==============================================================================
# 8. TOTAL CLIPS
# ==============================================================================

actual_total = int(
    results_df["metadata_rows"].sum()
) if len(results_df) else 0


expected_total = sum(
    EXPECTED_CLIPS.values()
)


print()
print(
    "Expected total clips :",
    f"{expected_total:,}"
)

print(
    "Metadata total clips :",
    f"{actual_total:,}"
)


total_pass = (
    actual_total
    == expected_total
)


print(
    "Total clip count     :",
    "PASS"
    if total_pass
    else "FAIL"
)


# ==============================================================================
# 9. FINAL RESULT
# ==============================================================================

print()
print("=" * 78)
print("CELL 9 FINAL RESULT")
print("=" * 78)


if global_pass and total_pass:

    print(
        "✅ CELL 9 — NPZ SHARD VALIDATION : PASS"
    )

    print()
    print(
        "All Phase 6 v4 NPZ shards are readable."
    )

    print(
        "Metadata → shard → shard_index mapping is valid."
    )

    print(
        "video001 + video002 + video003 are present."
    )

    print(
        f"Total clips : {actual_total:,}"
    )

else:

    print(
        "❌ CELL 9 — NPZ SHARD VALIDATION : FAIL"
    )

    print()
    print(
        "Review the errors above."
    )

    print(
        "No files were modified."
    )


print()
print("=" * 78)
print("CELL 9 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 10
# VERIFY GLOBAL CLIP MAPPING — PHASE 6 v4
# READ ONLY
# ==============================================================================

from pathlib import Path
import pandas as pd


print("=" * 78)
print("PHASE 6.5 — CELL 10")
print("VERIFY GLOBAL CLIP MAPPING — PHASE 6 v4")
print("=" * 78)


# ==============================================================================
# 1. CONFIGURATION
# ==============================================================================

PROJECT_ROOT = Path(
    r"C:\LipReadingSSL"
)

OUTPUT_ROOT = PROJECT_ROOT / "output"

VIDEO_IDS = [
    "video001",
    "video002",
    "video003",
]


EXPECTED_CLIPS = {
    "video001": 40076,
    "video002": 31662,
    "video003": 26273,
}


# ==============================================================================
# 2. PHASE 6 v4 SCHEMA
# ==============================================================================

REQUIRED_COLUMNS = [
    "clip_id",
    "video",
    "shard",
    "shard_index",
    "start_frame",
    "end_frame",
]


# ==============================================================================
# 3. LOAD METADATA
# ==============================================================================

PHASE65_METADATA = {}

for video_id in VIDEO_IDS:

    metadata_path = (
        OUTPUT_ROOT
        / video_id
        / "clip_metadata.csv"
    )

    print()
    print("-" * 78)
    print(video_id)
    print("-" * 78)

    print(
        "Metadata:",
        metadata_path
    )

    if not metadata_path.exists():

        raise FileNotFoundError(
            f"{video_id}: metadata not found:\n"
            f"{metadata_path}"
        )

    df = pd.read_csv(
        metadata_path
    )

    missing = [
        col
        for col in REQUIRED_COLUMNS
        if col not in df.columns
    ]

    if missing:

        raise RuntimeError(
            f"{video_id}: "
            f"Missing Phase 6 v4 columns: {missing}"
        )

    PHASE65_METADATA[video_id] = df

    print(
        "Rows:",
        f"{len(df):,}"
    )

    print(
        "Schema: PASS"
    )


# ==============================================================================
# 4. GLOBAL CLIP ID UNIQUENESS
# ==============================================================================

print()
print("=" * 78)
print("CHECK GLOBAL CLIP ID UNIQUENESS")
print("=" * 78)


all_clip_ids = []

for video_id in VIDEO_IDS:

    df = PHASE65_METADATA[
        video_id
    ]

    ids = (
        df["clip_id"]
        .astype(str)
        .str.strip()
    )

    all_clip_ids.extend(
        ids.tolist()
    )


total_clip_ids = len(
    all_clip_ids
)

unique_clip_ids = len(
    set(all_clip_ids)
)

duplicate_global = (
    total_clip_ids
    - unique_clip_ids
)


print(
    "Total clip IDs  :",
    f"{total_clip_ids:,}"
)

print(
    "Unique clip IDs :",
    f"{unique_clip_ids:,}"
)

print(
    "Duplicates      :",
    f"{duplicate_global:,}"
)


if duplicate_global:

    print(
        "❌ GLOBAL CLIP ID UNIQUENESS : FAIL"
    )

    global_mapping_pass = False

else:

    print(
        "✅ GLOBAL CLIP ID UNIQUENESS : PASS"
    )

    global_mapping_pass = True


# ==============================================================================
# 5. EXPECTED GLOBAL TOTAL
# ==============================================================================

expected_global_total = sum(
    EXPECTED_CLIPS.values()
)

print()
print("=" * 78)
print("GLOBAL CLIP COUNT")
print("=" * 78)

print(
    "Expected :",
    f"{expected_global_total:,}"
)

print(
    "Actual   :",
    f"{total_clip_ids:,}"
)


if total_clip_ids == expected_global_total:

    print(
        "✅ GLOBAL CLIP COUNT : PASS"
    )

else:

    print(
        "❌ GLOBAL CLIP COUNT : FAIL"
    )

    global_mapping_pass = False


# ==============================================================================
# 6. VERIFY EACH VIDEO
# ==============================================================================

print()
print("=" * 78)
print("VERIFY VIDEO-SPECIFIC GLOBAL MAPPING")
print("=" * 78)


video_results = []


for video_id in VIDEO_IDS:

    print()
    print("=" * 78)
    print(f"VIDEO : {video_id}")
    print("=" * 78)

    df = PHASE65_METADATA[
        video_id
    ].copy()


    # --------------------------------------------------------------------------
    # Expected count
    # --------------------------------------------------------------------------

    expected_count = (
        EXPECTED_CLIPS[video_id]
    )

    actual_count = len(df)

    count_ok = (
        actual_count
        == expected_count
    )

    print(
        "Clip count :",
        "PASS"
        if count_ok
        else
        (
            f"FAIL "
            f"(actual={actual_count:,}, "
            f"expected={expected_count:,})"
        )
    )


    if not count_ok:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Video consistency
    # --------------------------------------------------------------------------

    video_values = (
        df["video"]
        .astype(str)
        .str.strip()
    )

    wrong_video = int(
        (
            video_values
            != video_id
        ).sum()
    )

    print(
        "Video consistency :",
        "PASS"
        if wrong_video == 0
        else f"FAIL ({wrong_video:,})"
    )

    if wrong_video:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Clip ID uniqueness within video
    # --------------------------------------------------------------------------

    duplicate_clip_ids = int(
        df["clip_id"]
        .astype(str)
        .str.strip()
        .duplicated()
        .sum()
    )

    print(
        "Local clip ID uniqueness :",
        "PASS"
        if duplicate_clip_ids == 0
        else f"FAIL ({duplicate_clip_ids:,})"
    )

    if duplicate_clip_ids:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Expected clip ID range
    #
    # Phase 6 v4:
    #
    # video001_clip_000000
    # ...
    # video001_clip_040075
    #
    # --------------------------------------------------------------------------

    expected_prefix = (
        f"{video_id}_clip_"
    )

    clip_ids = (
        df["clip_id"]
        .astype(str)
        .str.strip()
    )


    wrong_prefix = int(
        (
            ~clip_ids.str.startswith(
                expected_prefix
            )
        ).sum()
    )


    print(
        "Clip ID prefix :",
        "PASS"
        if wrong_prefix == 0
        else f"FAIL ({wrong_prefix:,})"
    )


    if wrong_prefix:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Extract numeric clip index
    # --------------------------------------------------------------------------

    numeric_clip_index = pd.to_numeric(
        clip_ids.str.extract(
            r"_clip_(\d+)$",
            expand=False
        ),
        errors="coerce"
    )


    invalid_clip_index = int(
        numeric_clip_index.isna().sum()
    )


    print(
        "Clip ID numeric index :",
        "PASS"
        if invalid_clip_index == 0
        else f"FAIL ({invalid_clip_index:,})"
    )


    if invalid_clip_index:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Verify clip indices form complete range
    # --------------------------------------------------------------------------

    expected_indices = set(
        range(
            expected_count
        )
    )

    actual_indices = set(
        numeric_clip_index.dropna()
        .astype(int)
        .tolist()
    )


    missing_indices = sorted(
        expected_indices
        - actual_indices
    )

    extra_indices = sorted(
        actual_indices
        - expected_indices
    )


    print(
        "Missing clip indices :",
        f"{len(missing_indices):,}"
    )

    print(
        "Extra clip indices   :",
        f"{len(extra_indices):,}"
    )


    range_ok = (
        not missing_indices
        and not extra_indices
        and len(actual_indices)
        == expected_count
    )


    print(
        "Clip index continuity :",
        "PASS"
        if range_ok
        else "FAIL"
    )


    if not range_ok:

        global_mapping_pass = False


        if missing_indices:

            print(
                "First missing:",
                missing_indices[:10]
            )

        if extra_indices:

            print(
                "First extra:",
                extra_indices[:10]
            )


    # --------------------------------------------------------------------------
    # Verify metadata ordering
    #
    # The expected ordering is:
    #
    # clip_000000
    # clip_000001
    # clip_000002
    # ...
    # --------------------------------------------------------------------------

    ordering_ok = (
        numeric_clip_index
        .is_monotonic_increasing
    )


    print(
        "Clip ordering :",
        "PASS"
        if ordering_ok
        else "FAIL"
    )


    if not ordering_ok:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Verify shard_index is numeric
    # --------------------------------------------------------------------------

    shard_index = pd.to_numeric(
        df["shard_index"],
        errors="coerce"
    )


    invalid_shard_index = int(
        shard_index.isna().sum()
    )


    print(
        "Shard index validity :",
        "PASS"
        if invalid_shard_index == 0
        else f"FAIL ({invalid_shard_index:,})"
    )


    if invalid_shard_index:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Verify start/end frame
    # --------------------------------------------------------------------------

    start_frame = pd.to_numeric(
        df["start_frame"],
        errors="coerce"
    )

    end_frame = pd.to_numeric(
        df["end_frame"],
        errors="coerce"
    )


    invalid_frames = int(
        (
            start_frame.isna()
            |
            end_frame.isna()
        ).sum()
    )


    invalid_range = int(
        (
            end_frame
            <= start_frame
        ).sum()
    )


    print(
        "Frame values :",
        "PASS"
        if invalid_frames == 0
        else f"FAIL ({invalid_frames:,})"
    )

    print(
        "Frame ranges :",
        "PASS"
        if invalid_range == 0
        else f"FAIL ({invalid_range:,})"
    )


    if invalid_frames or invalid_range:

        global_mapping_pass = False


    # --------------------------------------------------------------------------
    # Store result
    # --------------------------------------------------------------------------

    video_pass = (
        count_ok
        and wrong_video == 0
        and duplicate_clip_ids == 0
        and wrong_prefix == 0
        and invalid_clip_index == 0
        and range_ok
        and ordering_ok
        and invalid_shard_index == 0
        and invalid_frames == 0
        and invalid_range == 0
    )


    video_results.append({
        "video_id": video_id,
        "actual_clips": actual_count,
        "expected_clips": expected_count,
        "duplicate_clip_ids": duplicate_clip_ids,
        "wrong_video": wrong_video,
        "wrong_prefix": wrong_prefix,
        "invalid_clip_index": invalid_clip_index,
        "missing_indices": len(missing_indices),
        "extra_indices": len(extra_indices),
        "invalid_shard_index": invalid_shard_index,
        "invalid_frames": invalid_frames,
        "invalid_frame_ranges": invalid_range,
        "status": (
            "PASS"
            if video_pass
            else "FAIL"
        ),
    })


# ==============================================================================
# 7. VERIFY GLOBAL CLIP ORDER ACROSS VIDEOS
# ==============================================================================

print()
print("=" * 78)
print("VERIFY GLOBAL VIDEO → CLIP ORDER")
print("=" * 78)


global_order_ok = True


for video_id in VIDEO_IDS:

    df = PHASE65_METADATA[
        video_id
    ]

    clip_ids = (
        df["clip_id"]
        .astype(str)
        .str.strip()
    )


    if len(clip_ids) == 0:

        global_order_ok = False

        print(
            video_id,
            ": ❌ EMPTY"
        )

        continue


    first_id = clip_ids.iloc[0]

    last_id = clip_ids.iloc[-1]


    expected_first = (
        f"{video_id}_clip_000000"
    )

    expected_last = (
        f"{video_id}_clip_{EXPECTED_CLIPS[video_id]-1:06d}"
    )


    first_ok = (
        first_id
        == expected_first
    )

    last_ok = (
        last_id
        == expected_last
    )


    print()
    print(video_id)

    print(
        "First:",
        first_id,
        "→",
        "PASS"
        if first_ok
        else f"FAIL (expected {expected_first})"
    )

    print(
        "Last :",
        last_id,
        "→",
        "PASS"
        if last_ok
        else f"FAIL (expected {expected_last})"
    )


    if not first_ok or not last_ok:

        global_order_ok = False


print()

if global_order_ok:

    print(
        "✅ GLOBAL VIDEO → CLIP ORDER : PASS"
    )

else:

    print(
        "❌ GLOBAL VIDEO → CLIP ORDER : FAIL"
    )

    global_mapping_pass = False


# ==============================================================================
# 8. FINAL RESULT TABLE
# ==============================================================================

print()
print("=" * 78)
print("CELL 10 RESULT TABLE")
print("=" * 78)


results_df = pd.DataFrame(
    video_results
)


if len(results_df):

    print(
        results_df.to_string(
            index=False
        )
    )


# ==============================================================================
# 9. FINAL
# ==============================================================================

print()
print("=" * 78)
print("CELL 10 FINAL RESULT")
print("=" * 78)


if global_mapping_pass:

    print(
        "✅ CELL 10 — GLOBAL CLIP MAPPING : PASS"
    )

    print()
    print(
        "Phase 6 v4 metadata is globally consistent."
    )

    print(
        "video001 + video002 + video003 are present."
    )

    print(
        f"Total clips : {total_clip_ids:,}"
    )

    print(
        "No files were modified."
    )

else:

    print(
        "❌ CELL 10 — GLOBAL CLIP MAPPING : FAIL"
    )

    print()
    print(
        "Review the failed checks above."
    )

    print(
        "No files were modified."
    )


print()
print("=" * 78)
print("CELL 10 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 18
# TRACE ACTUAL LABEL / TRANSCRIPT FILE I/O
# READ-ONLY
# ==============================================================================

from pathlib import Path
import json
import re

PROJECT_ROOT = Path(r"C:\LipReadingSSL")
NOTEBOOK_DIR = PROJECT_ROOT / "notebooks"

TARGET_NOTEBOOKS = [
    "phase6_clip_Generation.ipynb",
    "phase6.5_clip_source_alignment.ipynb",
    "phase7_dataset_builder.ipynb",
    "phase7.5_dataset_QA.ipynb",
]

PATTERNS = [
    r"\.read_csv\s*\(",
    r"pd\.read_csv\s*\(",
    r"\.read_table\s*\(",
    r"pd\.read_table\s*\(",
    r"\.read_json\s*\(",
    r"pd\.read_json\s*\(",
    r"\.to_csv\s*\(",
    r"\.to_json\s*\(",
    r"open\s*\(",
]

LABEL_TERMS = [
    "label",
    "labels",
    "annotation",
    "annot",
    "transcript",
    "sentence",
    "speaker",
    "normalized_text",
    "text",
    "start_frame",
    "end_frame",
    "clip_id",
    "dataset_id",
]

print("=" * 78)
print("PHASE 6.5 — CELL 18")
print("TRACE ACTUAL LABEL / TRANSCRIPT FILE I/O")
print("READ-ONLY")
print("=" * 78)

for notebook_name in TARGET_NOTEBOOKS:

    nb_path = NOTEBOOK_DIR / notebook_name

    print()
    print("=" * 78)
    print(f"NOTEBOOK : {notebook_name}")
    print("=" * 78)

    if not nb_path.exists():

        print("❌ Notebook not found")
        continue

    with open(
        nb_path,
        "r",
        encoding="utf-8"
    ) as f:

        notebook = json.load(f)

    found = 0

    for cell_index, cell in enumerate(
        notebook.get("cells", [])
    ):

        source = "".join(
            cell.get("source", [])
        )

        if not source.strip():
            continue

        lines = source.splitlines()

        for line_index, line in enumerate(lines):

            # --------------------------------------------------------------
            # Only inspect actual file I/O statements
            # --------------------------------------------------------------

            is_io = any(
                re.search(
                    pattern,
                    line,
                    flags=re.IGNORECASE
                )
                for pattern in PATTERNS
            )

            if not is_io:
                continue

            # --------------------------------------------------------------
            # Check whether nearby context contains label-related terms
            # --------------------------------------------------------------

            context_start = max(
                0,
                line_index - 5
            )

            context_end = min(
                len(lines),
                line_index + 6
            )

            context = "\n".join(
                lines[context_start:context_end]
            ).lower()

            related_terms = [
                term
                for term in LABEL_TERMS
                if term.lower() in context
            ]

            if not related_terms:
                continue

            found += 1

            print()
            print("-" * 78)
            print(
                f"CELL : {cell_index}"
            )
            print(
                f"LINE : {line_index + 1}"
            )
            print(
                "RELATED TERMS : "
                + ", ".join(
                    sorted(set(related_terms))
                )
            )
            print("-" * 78)

            for i in range(
                context_start,
                context_end
            ):

                marker = (
                    ">>> "
                    if i == line_index
                    else "    "
                )

                print(
                    f"{marker}{i + 1:4d}: "
                    f"{lines[i]}"
                )

    print()
    print(
        f"Relevant file-I/O blocks found : {found}"
    )

print()
print("=" * 78)
print("CELL 18 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 20
# FIND ALL CODE THAT WRITES CSV / LABEL / ANNOTATION OUTPUT
# READ-ONLY
# ==============================================================================

from pathlib import Path
import json
import re

PROJECT_ROOT = Path(
    r"C:\LipReadingSSL"
)

NOTEBOOK_DIR = PROJECT_ROOT / "notebooks"

print("=" * 78)
print("PHASE 6.5 — CELL 20")
print("FIND ALL CODE THAT WRITES CSV / LABEL / ANNOTATION OUTPUT")
print("READ-ONLY")
print("=" * 78)

NOTEBOOKS = sorted(
    NOTEBOOK_DIR.glob("*.ipynb")
)

# ----------------------------------------------------------------------
# Patterns that indicate file creation / writing
# ----------------------------------------------------------------------

WRITE_PATTERNS = [
    r"\.to_csv\s*\(",
    r"\.to_json\s*\(",
    r"\.to_excel\s*\(",
    r"open\s*\([^)]*,\s*[\"']w",
    r"open\s*\([^)]*,\s*[\"']a",
    r"json\.dump\s*\(",
    r"csv\.writer",
]

# ----------------------------------------------------------------------
# Terms strongly related to annotation / labeling
# ----------------------------------------------------------------------

LABEL_TERMS = [
    "label",
    "labels",
    "annotation",
    "annot",
    "transcript",
    "sentence",
    "speaker",
    "normalized_text",
    "text",
    "start_frame",
    "end_frame",
    "clip_id",
    "dataset_id",
]

total_matches = 0

for notebook_path in NOTEBOOKS:

    try:
        with open(
            notebook_path,
            "r",
            encoding="utf-8"
        ) as f:
            notebook = json.load(f)

    except Exception as e:

        print()
        print(
            f"⚠️ Cannot read: {notebook_path.name}"
        )
        print(e)
        continue

    notebook_matches = 0

    for cell_index, cell in enumerate(
        notebook.get("cells", [])
    ):

        source = "".join(
            cell.get("source", [])
        )

        if not source.strip():
            continue

        lines = source.splitlines()

        for line_index, line in enumerate(lines):

            # ----------------------------------------------------------
            # Is this a write operation?
            # ----------------------------------------------------------

            is_write = any(
                re.search(
                    pattern,
                    line,
                    flags=re.IGNORECASE
                )
                for pattern in WRITE_PATTERNS
            )

            if not is_write:
                continue

            # ----------------------------------------------------------
            # Inspect nearby context
            # ----------------------------------------------------------

            context_start = max(
                0,
                line_index - 8
            )

            context_end = min(
                len(lines),
                line_index + 9
            )

            context = "\n".join(
                lines[
                    context_start:context_end
                ]
            ).lower()

            related_terms = [
                term
                for term in LABEL_TERMS
                if term.lower() in context
            ]

            # ----------------------------------------------------------
            # Only show writes related to labeling/data metadata
            # ----------------------------------------------------------

            if not related_terms:
                continue

            total_matches += 1
            notebook_matches += 1

            print()
            print("=" * 78)
            print(
                f"NOTEBOOK : {notebook_path.name}"
            )
            print(
                f"CELL     : {cell_index}"
            )
            print(
                f"LINE     : {line_index + 1}"
            )
            print(
                "RELATED  : "
                + ", ".join(
                    sorted(set(related_terms))
                )
            )
            print("=" * 78)

            for i in range(
                context_start,
                context_end
            ):

                marker = (
                    ">>> "
                    if i == line_index
                    else "    "
                )

                print(
                    f"{marker}{i + 1:4d}: "
                    f"{lines[i]}"
                )

    print()
    print(
        f"{notebook_path.name} "
        f"→ {notebook_matches} relevant write blocks"
    )

print()
print("=" * 78)
print(
    f"TOTAL RELEVANT WRITE BLOCKS : {total_matches}"
)
print("=" * 78)

print()
print("CELL 20 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 30
# CHECK DATASET_MASTER ANNOTATION COMPLETENESS
# READ-ONLY
# ==============================================================================

from pathlib import Path
import pandas as pd

DATASET_MASTER_PATH = Path(
    r"C:\LipReadingSSL\output\dataset\dataset_master.csv"
)

print("=" * 78)
print("PHASE 6.5 — CELL 30")
print("CHECK DATASET_MASTER ANNOTATION COMPLETENESS")
print("READ-ONLY")
print("=" * 78)

# ------------------------------------------------------------------------------
# 1. Load
# ------------------------------------------------------------------------------

if not DATASET_MASTER_PATH.exists():
    raise FileNotFoundError(
        f"dataset_master.csv not found:\n{DATASET_MASTER_PATH}"
    )

df = pd.read_csv(
    DATASET_MASTER_PATH,
    low_memory=False
)

print()
print(f"Rows    : {len(df):,}")
print(f"Columns : {len(df.columns)}")

# ------------------------------------------------------------------------------
# 2. Annotation columns
# ------------------------------------------------------------------------------

ANNOTATION_COLUMNS = [
    "speaker_id",
    "sentence_id",
    "transcript",
    "normalized_text",
    "language",
    "verified",
]

missing_columns = [
    column
    for column in ANNOTATION_COLUMNS
    if column not in df.columns
]

if missing_columns:
    print()
    print("Missing annotation columns:")
    for column in missing_columns:
        print(f" - {column}")

    raise RuntimeError(
        "Required annotation columns are missing."
    )

# ------------------------------------------------------------------------------
# 3. Completeness
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("ANNOTATION COMPLETENESS")
print("=" * 78)

for column in ANNOTATION_COLUMNS:

    series = df[column]

    missing_mask = (
        series.isna()
        | series.astype(str).str.strip().isin(
            ["", "nan", "None", "NULL", "UNKNOWN"]
        )
    )

    missing_count = int(missing_mask.sum())
    valid_count = int(len(df) - missing_count)

    percentage = (
        valid_count / len(df) * 100
        if len(df)
        else 0
    )

    print()
    print(f"{column}")
    print(f"  Valid   : {valid_count:,}")
    print(f"  Missing : {missing_count:,}")
    print(f"  Valid % : {percentage:.2f}%")

# ------------------------------------------------------------------------------
# 4. Video distribution
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("ANNOTATION COVERAGE BY VIDEO")
print("=" * 78)

for video_id, group in df.groupby(
    "video_id",
    dropna=False
):

    print()
    print(f"{video_id}")

    for column in ANNOTATION_COLUMNS:

        series = group[column]

        missing_mask = (
            series.isna()
            | series.astype(str).str.strip().isin(
                ["", "nan", "None", "NULL", "UNKNOWN"]
            )
        )

        valid_count = int(
            (~missing_mask).sum()
        )

        print(
            f"  {column:<20} "
            f"{valid_count:,}/{len(group):,}"
        )

# ------------------------------------------------------------------------------
# 5. Transcript examples
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("TRANSCRIPT SAMPLE")
print("=" * 78)

sample_columns = [
    "video_id",
    "clip_id",
    "speaker_id",
    "sentence_id",
    "transcript",
    "normalized_text",
]

print(
    df[sample_columns]
    .head(20)
    .to_string(index=False)
)

# ------------------------------------------------------------------------------
# 6. Final
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("CELL 30 COMPLETE — READ ONLY")
print("=" * 78)
```

```python
# ==============================================================================
# PHASE 6.5 — CELL 31
# INSPECT EXISTING LABEL_QA ANNOTATION VALUES
# READ-ONLY
# ==============================================================================

from pathlib import Path
import pandas as pd

LABEL_QA_PATH = Path(
    r"C:\LipReadingSSL\output\dataset\analysis\label_qa.csv"
)

print("=" * 78)
print("PHASE 6.5 — CELL 31")
print("INSPECT EXISTING LABEL_QA ANNOTATION VALUES")
print("READ-ONLY")
print("=" * 78)

print()
print("Path   :", LABEL_QA_PATH)
print("Exists :", LABEL_QA_PATH.exists())

if not LABEL_QA_PATH.exists():
    raise FileNotFoundError(
        f"label_qa.csv not found:\n{LABEL_QA_PATH}"
    )

# ------------------------------------------------------------------------------
# Load
# ------------------------------------------------------------------------------

df = pd.read_csv(
    LABEL_QA_PATH,
    low_memory=False
)

print()
print("=" * 78)
print("BASIC INFORMATION")
print("=" * 78)

print(
    f"Rows    : {len(df):,}"
)

print(
    f"Columns : {len(df.columns)}"
)

# ------------------------------------------------------------------------------
# Columns
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("COLUMNS")
print("=" * 78)

for i, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i:02d}. {column}"
    )

# ------------------------------------------------------------------------------
# Annotation fields
# ------------------------------------------------------------------------------

annotation_columns = [
    column
    for column in [
        "video_id",
        "clip_id",
        "speaker_id",
        "sentence_id",
        "transcript",
        "normalized_text",
        "language",
        "verified",
    ]
    if column in df.columns
]

print()
print("=" * 78)
print("ANNOTATION FIELD STATUS")
print("=" * 78)

for column in annotation_columns:

    series = df[column]

    missing = (
        series.isna()
        | series.astype(str).str.strip().isin(
            [
                "",
                "nan",
                "None",
                "NULL",
                "UNKNOWN",
            ]
        )
    )

    valid_count = int(
        (~missing).sum()
    )

    missing_count = int(
        missing.sum()
    )

    unique_count = int(
        series[~missing].nunique()
    )

    print()
    print(column)
    print(
        f"  Valid       : {valid_count:,}"
    )
    print(
        f"  Missing     : {missing_count:,}"
    )
    print(
        f"  Unique      : {unique_count:,}"
    )

# ------------------------------------------------------------------------------
# Video distribution
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("VIDEO DISTRIBUTION")
print("=" * 78)

if "video_id" in df.columns:

    counts = (
        df["video_id"]
        .value_counts(dropna=False)
        .sort_index()
    )

    for video_id, count in counts.items():

        print(
            f"{str(video_id):<15} "
            f"{int(count):,} rows"
        )

# ------------------------------------------------------------------------------
# Transcript examples
# ------------------------------------------------------------------------------

if "transcript" in df.columns:

    print()
    print("=" * 78)
    print("NON-EMPTY TRANSCRIPT SAMPLE")
    print("=" * 78)

    transcript_mask = (
        df["transcript"].notna()
        & (
            df["transcript"]
            .astype(str)
            .str.strip()
            .ne("")
        )
    )

    sample_columns = [
        column
        for column in [
            "video_id",
            "clip_id",
            "speaker_id",
            "sentence_id",
            "transcript",
            "normalized_text",
            "language",
            "verified",
        ]
        if column in df.columns
    ]

    sample = (
        df.loc[
            transcript_mask,
            sample_columns
        ]
        .head(20)
    )

    if len(sample):

        print(
            sample.to_string(
                index=False
            )
        )

    else:

        print(
            "No non-empty transcript found."
        )

# ------------------------------------------------------------------------------
# Unique transcript sample
# ------------------------------------------------------------------------------

if "transcript" in df.columns:

    print()
    print("=" * 78)
    print("UNIQUE TRANSCRIPT VALUES")
    print("=" * 78)

    values = (
        df["transcript"]
        .dropna()
        .astype(str)
        .str.strip()
    )

    values = values[
        values.ne("")
    ]

    unique_values = (
        values
        .drop_duplicates()
        .head(50)
    )

    if len(unique_values):

        for i, value in enumerate(
            unique_values,
            start=1
        ):

            print(
                f"{i:02d}. {value}"
            )

    else:

        print(
            "No transcript values found."
        )

# ------------------------------------------------------------------------------
# Final
# ------------------------------------------------------------------------------

print()
print("=" * 78)
print("CELL 31 COMPLETE — READ ONLY")
print("=" * 78)
```
